## FRED (Federal Reserve Economic Data): Free public db
https://fred.stlouisfed.org/

```python
cpi = pd.read_csv('https://fred.stlouisfed.org/graph/fredgraph.csv?id=CPIAUCSL')
cpi.columns = ['date','cpi']
# print(cpi.describe())
cpi['year'] = pd.to_datetime(cpi['date']).dt.year
annual = cpi.groupby('year')['cpi'].mean().reset_index()
base = annual.loc[annual.year==2025,'cpi'].values[0]
annual['deflator'] = base / annual['cpi']
df = df.merge(annual[['year','deflator']], left_on='ReleaseYear', right_on='year', how='left') # only works once, reset to rerun
df['RealPrice'] = df['Price'] * df['deflator']
```

```python
import requests, io
url = 'https://en.wikipedia.org/wiki/List_of_best-selling_PC_games'
response = requests.get(url, headers={'User-Agent': 'Mozilla/5.0 (OPIM 5641 student project)'}) 
# print(response.status_code) # 200, nice
tables = pd.read_html(io.StringIO(response.text))
# for i, t in enumerate(tables): print(i, list(t.columns)[:4]) # table where
wiki = tables[1]
wiki['WikiSalesM'] = wiki['Total copies sold'].str.extract(r'([\d.]+)').astype(float)

# clean and parse

def clean(s):
    return (s.str.replace('™','', regex=False).str.replace('®','', regex=False)
            .str.strip().str.lower())
tmp = df[['Name','TotalReviews']].copy()
tmp['key'] = clean(tmp['Name'])
wiki['key'] = clean(wiki['Game'].astype(str))
# print('uhh: ', wiki)# nice

# parse merge and match and join n allat
foundMatch = tmp.merge(wiki[['key','WikiSalesM']], on='key', how='inner')
foundMatch['EstUnitsM'] = foundMatch['TotalReviews'] * 30 / 1e6   # boxleiter estimate
foundMatch['ratio'] = foundMatch['EstUnitsM'] / foundMatch['WikiSalesM']
print(len(foundMatch), 'titles foundMatch') # sick
foundMatch.sort_values('WikiSalesM', ascending=False).head(15)
```

## Business Framing Decision:
### A small publisher decides to fund 6 premium indie titles at 150k each, totaling 900k. 

## **How likely is it for a given slate to fail to earn it back?**

```python
slate_pool = df[ df['Genres'].fillna('').str.contains('Indie') & (df['Price'] >= 10) & df['ReleaseYear'].between(2020, 2024) ].copy()
# indie only
# exclude f2p
# exclude 2025, focus on when market was already crowded
print(len(slate_pool)) # how many non f2p indie titles were there between 2020 and 2024
```

## Assumptions

Big thing: Steam publishes review counts, Steam doesn't publish sales. 
As a result, to approximate a titles lifetime net revenue: 
reviews * 30 * game price * 70%

30: the Boxleiter multiplier: converts reviews to copies sold. GameDiscoverCo's survey had a median of 30 for recent releases as of 2021

game price: uses price at scrape time, which is often a rough estimate for avg selling price as opposed to launch price

70%: steam takes 30%, cest la vie. revenue then equals 70% of (gross dollars)

```python
slate_pool['NetRev'] = slate_pool['TotalReviews'] * 30 * slate_pool['Price'] * 0.70
slate_pool['NetRev'].describe()
```

# Simulate

```python
np.random.seed(5641) #yep
n_sims, n_titles, budget = 10_000, 6, 900_000

revenues = []
profits = []
for sim in range(n_sims):
    slate = slate_pool['NetRev'].sample(n=n_titles, replace=True).values  # fund 6 random titles
    revenues.append(slate.sum())
    profits.append(slate.sum() - budget) # rev - spend

revenues = np.array(revenues)         # arrays for math reasons
profits = np.array(profits)
```

```python
p5, p50, p95 = np.percentile(revenues, [5, 50, 95])   # the percentile lines the assignment asks for

plt.figure(figsize=(10, 4))
plt.hist(np.log10(revenues), bins=60, color='steelblue')   # log10 my beloved

for value, name, color in [(p5, '5th', 'gray'), (p50, 'median', 'red'), (p95, '95th', 'gray')]: plt.axvline(np.log10(value), color=color, linestyle='dashed', label=f'{name}: ${value:,.0f}')
plt.axvline(np.log10(budget), color='black', linewidth=2, label=f'break-even: ${budget:,.0f}')

plt.legend()
plt.title(f'10,000 simulated slates — P(loss) = {(profits < 0).mean():.1%}')
plt.xlabel('log10(slate net revenue $)   ·   4 = $10k · 5 = $100k · 6 = $1M · 7 = $10M')
plt.ylabel('number of slates');
```
